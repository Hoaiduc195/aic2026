```python
import pandas as pd
import numpy as np
from PIL import Image
from pathlib import Path

# Create mock keyframe data to test the pipeline
test_dir = Path('tests/mock_visual_embedding')
test_dir.mkdir(parents=True, exist_ok=True)
img_path = test_dir / 'test_frame.jpg'
Image.fromarray(np.zeros((224, 224, 3), dtype=np.uint8)).save(img_path)

df = pd.DataFrame([{
    'video_id': 'vid123',
    'original_frame_id': 0,
    'timestamp_ms': 1000,
    'storage_uri': f'file://{img_path.absolute()}',
    'retrieval_roles': ['shot_anchor'],
    'quality_scores': {'brightness_score': 0.5, 'blur_score': 0.1, 'contrast_score': 0.5, 'entropy_score': 0.8},
    'quality_route': 'retrieval_embedding',
    'selected_for_retrieval': True,
    'dataset_id': 'ds1',
    'dataset_version': 'v1'
}])
df.to_parquet(test_dir / 'vid123.parquet')
print('Mock data created.')
```
